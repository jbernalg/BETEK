1. Calculadora de Edad (Calentamiento) Escribe un programa que pregunte al usuario 
su año de nacimiento y el año actual. El programa debe calcular su edad y mostrar el 
siguiente mensaje en pantalla: "Tienes [edad] años o los cumplirás este año". 

```python
year_nacimiento = int(input('Ingrese el año de nacimiento: '))
year_actual = int(input('Ingrese el año actual: '))
edad_actual = year_actual - year_nacimiento

print(f'Tienes {edad_actual} años o los cumpliras este año')
```

2. Promedio de Satisfacción al Cliente (CX) Imagina que estás procesando los resultados de un formulario de encuestas de servicio al cliente. El programa debe pedir al usuario que ingrese las calificaciones de 3 encuestas diferentes (números 
enteros del 1 al 5). Calcula el promedio exacto de estas tres calificaciones (con decimales) y muéstralo en pantalla. 

```python
calificacion_encuesta1 = int(input('Ingrese la calificacion de la encuesta 1 (un numero del 1 al 5): '))
calificacion_encuesta2 = int(input('Ingrese la calificacion de la encuesta 2 (un numero del 1 al 5): '))
calificacion_encuesta3 = int(input('Ingrese la calificacion de la encuesta 3 (un numero del 1 al 5): '))

promedio_calificacion = (calificacion_encuesta1 + calificacion_encuesta2 + calificacion_encuesta3)/3
print(promedio_calificacion)
```

3. Auditoría de Tiempos de Ejecución Tienes un script automatizado que se ejecuta en el servidor y tarda cierta cantidad de minutos en procesar la data. Construye un programa que pida al usuario ingresar el total de minutos que tardó el proceso. 
Utilizando los operadores de división entera (//) y módulo (%), convierte ese tiempo a horas y minutos. Ejemplo: Si el usuario ingresa 130, el programa debe imprimir: "El proceso tardó 2 horas y 10 minutos". 

```python
total_minutos = int(input('Ingrese el total de minutos que tardo el proceso: '))
tiempo_convertido_h = (total_minutos // 60)
tiempo_convertido_min = total_minutos % 60

print(f'El proceso tardo {tiempo_convertido_h} horas y {tiempo_convertido_min} minutos')
```

4. Generador de Correos Corporativos Estás ayudando a Recursos Humanos a limpiar la base de datos de los nuevos empleados. Crea un programa que solicite al usuario: 

Su primer nombre.

Su primer apellido. 

El nombre de la empresa.  

El programa debe concatenar (unir) estos textos para imprimir el correo electrónico 
oficial. Ejemplo de salida: "Tu correo asignado es: nombre.apellido@empresa.com" 

```python
primer_nombre = input('Ingrese el primer nombre: ')
primer_apellido = input('Ingrese el primer apellido: ')
nombre_empresa = input('Ingrese el nombre de la empresa: ')

print(f'Tu correo asignado es: {primer_nombre}.{primer_apellido}@{nombre_empresa}.com')
```
